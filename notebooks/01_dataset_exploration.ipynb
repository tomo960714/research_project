```python
from pathlib import Path

from ecg_prediction.dataset import (
    PTBXLDataset,
    load_metadata,
    split_metadata
)
```

```python
data_dir = Path("../data/ptb-xl")
```

```python
metadata = load_metadata(data_dir)
```

```python
print(metadata.shape)
print(metadata.head())
print(metadata["sex"].value_counts())
```

```python
# check out split of the dataset
train_df = split_metadata(metadata, "train")
valid_df = split_metadata(metadata, "validation")
test_df = split_metadata(metadata, "test")

print("Training:", len(train_df))
print("Validation:", len(valid_df))
print("Testing:", len(test_df))
```

```python
# verify that the splits are disjoint
train_patients = set(train_df["patient_id"])
valid_patients = set(valid_df["patient_id"])
test_patients = set(test_df["patient_id"])

assert train_patients.isdisjoint(valid_patients)
assert train_patients.isdisjoint(test_patients)
assert valid_patients.isdisjoint(test_patients)

print("No patient overlap detected.")
```

```python
# load a sample of the dataset
train_dataset = PTBXLDataset(data_dir, train_df)

ecg, label = train_dataset[0]

print("ECG shape:", ecg.shape)
print("Label:", label.item())
print("ECG dtype:", ecg.dtype)
```

```python
# Try dataloader
from torch.utils.data import DataLoader

train_loader = DataLoader(
    train_dataset,
    batch_size=32,
    shuffle=True,
    num_workers=0,
)

ecg_batch, labels = next(iter(train_loader))

print(ecg_batch.shape)
print(labels.shape)
```
